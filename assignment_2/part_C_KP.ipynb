## B.3 Fairness criteria: Independence - selections rate parity (demographic parity)
Predictions should be statistically independent of a protected attribute and model should give positive predictions at the same rate across groups.

```python
def independence_parity(y_pred_binary, sensitive):
  results = {}
  for group in sorted(sensitive.unique()):
      mask = sensitive == group
      results[group] = y_pred_binary[mask].mean()
  return pd.Series(results)


def plot_independence_parity(feature_name, y_test, y_pred_rf, y_pred_lr, threshold_rf=0.5, threshold_lr=0.5):
  sensitive = data[feature_name].loc[y_test.index]

  ind_rf = independence_parity(y_pred_rf > threshold_rf, sensitive)
  ind_lr = independence_parity(y_pred_lr > threshold_lr, sensitive)

  pd.DataFrame({
      "Random Forest": ind_rf,
      "Logistic Regression": ind_lr
  }).plot(kind="bar", figsize=(10, 4))
  plt.title(f"Independence (Demographic Parity) by {feature_name}")
  plt.xlabel(f"{feature_name} group")
  plt.ylabel("Positive prediction rate")
  plt.legend()
  plt.tight_layout()
  plt.show()
```

```python
for fairness_var in fairness_vars:
    plot_independence_parity(fairness_var, y_test, y_pred, y_pred_lr)
```

## B.4 Fairness criteria: Separation - error rate parity
Error rate parity (also called separation) is a fairness criterion that focuses on equalizing error rates across groups, conditional on the true outcome.

The model should be equally accurate (or equally wrong) across groups.

```python
from sklearn.metrics import confusion_matrix

def separation_parity(y_true, y_pred_binary, sensitive):
  results = {}
  for group in sorted(sensitive.unique()):
      mask = sensitive == group
      tn, fp, fn, tp = confusion_matrix(y_true[mask], y_pred_binary[mask], labels=[0, 1]).ravel()
      results[group] = {
          "TPR": tp / (tp + fn) if (tp + fn) > 0 else 0,
          "FPR": fp / (fp + tn) if (fp + tn) > 0 else 0,
      }
  return pd.DataFrame(results).T


def plot_separation_parity(feature_name, y_test, y_pred_rf, y_pred_lr, threshold_rf=0.5, threshold_lr=0.5):
  sensitive = data[feature_name].loc[y_test.index]

  sep_rf = separation_parity(y_test, y_pred_rf > threshold_rf, sensitive)
  sep_lr = separation_parity(y_test, y_pred_lr > threshold_lr, sensitive)

  fig, axes = plt.subplots(1, 2, figsize=(14, 4), sharey=False)

  for ax, metric in zip(axes, ["TPR", "FPR"]):
      pd.DataFrame({
          "Random Forest": sep_rf[metric],
          "Logistic Regression": sep_lr[metric]
      }).plot(kind="bar", ax=ax)
      ax.set_title(f"{metric} by {feature_name}")
      ax.set_xlabel(f"{feature_name} group")
      ax.set_ylabel(metric)
      ax.legend()

  plt.suptitle(f"Separation by {feature_name}")
  plt.tight_layout()
  plt.show()
```

```python
for fairness_var in fairness_vars:
    plot_separation_parity(fairness_var, y_test, y_pred, y_pred_lr)
```

## B.5 Fairness criteria: Sufficiency
In binary predictions sufficiency reduces to precision parity across groups.

In our case precision = of everyone the model predicted as >75k, what fraction actually earns >75k

```python
def sufficiency_parity(y_true, y_pred_binary, sensitive):
  results = {}
  for group in sorted(sensitive.unique()):
      mask = sensitive == group
      tn, fp, fn, tp = confusion_matrix(y_true[mask], y_pred_binary[mask], labels=[0, 1]).ravel()
      results[group] = {
          "Precision": tp / (tp + fp) if (tp + fp) > 0 else 0,
      }
  return pd.DataFrame(results).T


def plot_sufficiency_parity(feature_name, y_test, y_pred_rf, y_pred_lr, threshold_rf=0.5, threshold_lr=0.5):
  sensitive = data[feature_name].loc[y_test.index]

  suf_rf = sufficiency_parity(y_test, y_pred_rf > threshold_rf, sensitive)
  suf_lr = sufficiency_parity(y_test, y_pred_lr > threshold_lr, sensitive)

  pd.DataFrame({
      "Random Forest": suf_rf["Precision"],
      "Logistic Regression": suf_lr["Precision"]
  }).plot(kind="bar", figsize=(10, 4))
  plt.title(f"Sufficiency (Precision) by {feature_name}")
  plt.xlabel(f"{feature_name} group")
  plt.ylabel("Precision")
  plt.legend()
  plt.tight_layout()
  plt.show()
```

```python
for fairness_var in fairness_vars:
    plot_sufficiency_parity(fairness_var, y_test, y_pred, y_pred_lr)
```

# Part C – Individual Fairness

In Task C, we aimed to improve **individual fairness** by applying two fairness-enhancing techniques: removing sensitive attributes from the model and applying *k-nearest-neighbour (k-NN) smoothing* to the predictions.

## Technique 1: Removing Sensitive Attributes

The first technique removes protected characteristics such as gender, age group, visible minority status, religion, immigration status, place of birth, and citizenship from the feature set. The goal of this approach is to reduce the possibility that individuals receive different predictions purely because they belong to different protected groups.

After removing these sensitive variables, the models still maintained relatively good predictive performance. This suggests that the prediction of high income does not fully depend on sensitive attributes alone. However, removing sensitive variables does not completely eliminate fairness concerns, since other variables in the dataset may still indirectly capture similar information and therefore act as proxies for protected characteristics.

## Technique 2: k-Nearest-Neighbour Smoothing

The second technique applies *k-nearest-neighbour smoothing* to the model predictions. Instead of relying entirely on the prediction generated for a single individual, the prediction is partially adjusted based on the predictions of similar individuals in the dataset.

This method follows the principle of **individual fairness**, which states that similar individuals should receive similar outcomes. By averaging predictions across neighbouring observations with comparable socioeconomic characteristics, the model produces more stable and consistent predictions.

The results show that after smoothing, the predictions became more similar across comparable individuals, improving the overall level of individual fairness.

## Trade-off Between Fairness and Performance

The findings highlight an important trade-off between fairness and predictive performance. While removing sensitive variables and applying smoothing techniques improves fairness, these adjustments may slightly reduce metrics such as accuracy, precision, or recall.

This trade-off is expected because the model becomes less focused on maximizing predictive performance and more focused on ensuring fair and consistent treatment across individuals. In the context of loan approval prediction, accepting a small reduction in performance may be reasonable if it helps reduce unfair differences between individuals and creates a model that is more ethically and socially defensible.

```python
# Task C - Individual Fairness

# In Task C we try to improve individual fairness.
# Individual fairness means that similar individuals should receive similar predictions.

from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import accuracy_score, balanced_accuracy_score, roc_auc_score
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
# C.1 Technique 1: Remove sensitive attributes

sensitive_features = [
    'Gender', 'VISMIN', 'ABOID', 'AGEGRP', 'RELIGION_DER', 'MARSTH',
    'ETHDER', 'IMMSTAT', 'GENSTAT', 'POB', 'POBPAR1', 'POBPAR2', 'Citizen'
]

X_fair = X.drop(columns=sensitive_features)

X_train_fair, X_test_fair, y_train_fair, y_test_fair, A_train_fair, A_test_fair = train_test_split(
    X_fair, y, AGEGRP, random_state=42, stratify=y
)
```

```python
# Random Forest without sensitive attributes

rf_fair = RandomForestClassifier(
    n_estimators=400,
    max_depth=15,
    min_samples_leaf=30,
    max_features=3,
    class_weight='balanced',
    random_state=42
)

rf_fair.fit(X_train_fair, y_train_fair)

y_pred_rf_fair = rf_fair.predict_proba(X_test_fair)[:, 1]
```

```python
# Logistic Regression without sensitive attributes

scaler_fair = StandardScaler()

X_train_fair_scaled = scaler_fair.fit_transform(X_train_fair)
X_test_fair_scaled = scaler_fair.transform(X_test_fair)

lr_fair = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

lr_fair.fit(X_train_fair_scaled, y_train_fair)

y_pred_lr_fair = lr_fair.predict_proba(X_test_fair_scaled)[:, 1]
```

```python
# Compare original models with models without sensitive attributes

threshold_rf = 0.5
threshold_lr = 0.45

results_task_c_1 = pd.DataFrame({
    "Model": [
        "Original Random Forest",
        "Fair Random Forest",
        "Original Logistic Regression",
        "Fair Logistic Regression"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred > threshold_rf),
        accuracy_score(y_test_fair, y_pred_rf_fair > threshold_rf),
        accuracy_score(y_test, y_pred_lr > threshold_lr),
        accuracy_score(y_test_fair, y_pred_lr_fair > threshold_lr)
    ],
    "Balanced Accuracy": [
        balanced_accuracy_score(y_test, y_pred > threshold_rf),
        balanced_accuracy_score(y_test_fair, y_pred_rf_fair > threshold_rf),
        balanced_accuracy_score(y_test, y_pred_lr > threshold_lr),
        balanced_accuracy_score(y_test_fair, y_pred_lr_fair > threshold_lr)
    ],
    "ROC AUC": [
        roc_auc_score(y_test, y_pred),
        roc_auc_score(y_test_fair, y_pred_rf_fair),
        roc_auc_score(y_test, y_pred_lr),
        roc_auc_score(y_test_fair, y_pred_lr_fair)
    ]
})

results_task_c_1
```

```python
# C.2 Technique 2: k-nearest-neighbour smoothing

# This technique smooths predictions by averaging the prediction of each individual
# with the predictions of similar individuals.

def knn_smoothing(X_values, predictions, k=10):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_values)

    neighbors = NearestNeighbors(n_neighbors=k)
    neighbors.fit(X_scaled)

    neighbor_indices = neighbors.kneighbors(X_scaled, return_distance=False)

    smoothed_predictions = []

    for indices in neighbor_indices:
        smoothed_predictions.append(predictions[indices].mean())

    return np.array(smoothed_predictions)
```

```python
# Apply smoothing to the fair models

y_pred_rf_smoothed = knn_smoothing(X_test_fair, y_pred_rf_fair, k=10)
y_pred_lr_smoothed = knn_smoothing(X_test_fair, y_pred_lr_fair, k=10)
```

```python
# Compare performance before and after smoothing

results_task_c_2 = pd.DataFrame({
    "Model": [
        "Fair Random Forest",
        "Smoothed Fair Random Forest",
        "Fair Logistic Regression",
        "Smoothed Fair Logistic Regression"
    ],
    "Accuracy": [
        accuracy_score(y_test_fair, y_pred_rf_fair > threshold_rf),
        accuracy_score(y_test_fair, y_pred_rf_smoothed > threshold_rf),
        accuracy_score(y_test_fair, y_pred_lr_fair > threshold_lr),
        accuracy_score(y_test_fair, y_pred_lr_smoothed > threshold_lr)
    ],
    "Balanced Accuracy": [
        balanced_accuracy_score(y_test_fair, y_pred_rf_fair > threshold_rf),
        balanced_accuracy_score(y_test_fair, y_pred_rf_smoothed > threshold_rf),
        balanced_accuracy_score(y_test_fair, y_pred_lr_fair > threshold_lr),
        balanced_accuracy_score(y_test_fair, y_pred_lr_smoothed > threshold_lr)
    ],
    "ROC AUC": [
        roc_auc_score(y_test_fair, y_pred_rf_fair),
        roc_auc_score(y_test_fair, y_pred_rf_smoothed),
        roc_auc_score(y_test_fair, y_pred_lr_fair),
        roc_auc_score(y_test_fair, y_pred_lr_smoothed)
    ]
})

results_task_c_2
```

```python
# Individual fairness metric:
# We measure how different the predictions are between similar individuals.
# A lower score means better individual fairness.

def individual_fairness_score(X_values, predictions, k=10):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_values)

    neighbors = NearestNeighbors(n_neighbors=k)
    neighbors.fit(X_scaled)

    neighbor_indices = neighbors.kneighbors(X_scaled, return_distance=False)

    differences = []

    for i, indices in enumerate(neighbor_indices):
        for j in indices:
            differences.append(abs(predictions[i] - predictions[j]))

    return np.mean(differences)
```

```python
individual_fairness_results = pd.DataFrame({
    "Model": [
        "Original Random Forest",
        "Fair Random Forest",
        "Smoothed Fair Random Forest",
        "Original Logistic Regression",
        "Fair Logistic Regression",
        "Smoothed Fair Logistic Regression"
    ],
    "Individual Fairness Score": [
        individual_fairness_score(X_test, y_pred),
        individual_fairness_score(X_test_fair, y_pred_rf_fair),
        individual_fairness_score(X_test_fair, y_pred_rf_smoothed),
        individual_fairness_score(X_test, y_pred_lr),
        individual_fairness_score(X_test_fair, y_pred_lr_fair),
        individual_fairness_score(X_test_fair, y_pred_lr_smoothed)
    ]
})

individual_fairness_results
```

```python
# Visualize individual fairness scores

individual_fairness_results.plot(
    x="Model",
    y="Individual Fairness Score",
    kind="bar",
    figsize=(10, 4),
    legend=False
)

plt.title("Individual Fairness Score by Model")
plt.ylabel("Similar mean prediction difference")
plt.xlabel("Model")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
